ANS.  Power BI is a business intelligence (BI) and data visualization platform developed by Microsoft. Q2.  Compare the following Power BI visuals:

● Bar Chart vs Column Chart

When would you prefer one over the other? Give one example for each pair

ANS.  Pie Chart vs Donut Chart

Pie Chart

A full circle divided into slices
Shows part-to-whole relationships
Labels are usually placed outside or on slices
Best for simple datasets (≤5 categories)

Donut Chart

Similar to a pie chart but with a hole in the center
Can display additional information in the center (like total value)
Slightly more modern and visually clean

When to prefer

Use Pie Chart when:
You want a straightforward, traditional view
The focus is purely on proportions

Use Donut Chart when:
You want to display totals or KPIs in the center
You prefer a cleaner dashboard design

Example

Pie Chart: Showing percentage share of sales by product category (Electronics, Clothing, Food)

Donut Chart: Same data, but with total sales displayed in the center

📊 Bar Chart vs Column Chart

Bar Chart

Horizontal bars
Categories on the Y-axis, values on the X-axis
Better for long category names
Column Chart
Vertical bars

Categories on the X-axis, values on the Y-axis
Commonly used for time-based data

When to prefer

Use Bar Chart when:
You have many categories
Category names are long
You want easy comparison across items

Use Column Chart when:
You are showing trends over time (months, years)
Categories are few and short
You want a standard vertical comparison

Example

Bar Chart: Comparing sales across different cities with long names

Column Chart: Showing monthly revenue trend (Jan–Dec)

✔️ Quick Summary

Pie vs Donut → Both show proportions; donut adds extra info in the center

Bar vs Column → Same concept rotated; choose based on readability and data type

Q3.  Explain the significance of:

● Star schema vs Snowflake schema

● Primary key vs Foreign key in relationships (Power BI)

Why is cardinality important?

ANS.  In Power BI, how you structure your data (modeling) is just as important as how you visualize it. A well-designed model ensures your reports are fast, accurate, and easy to maintain.

1. Star Schema vs. Snowflake Schema
These are two ways of organizing Fact tables (quantitative data like sales) and Dimension tables (descriptive data like product names or dates).

Star Schema (The Gold Standard): * Structure: A central Fact table connected directly to multiple Dimension tables. It looks like a star.

Significance: It is the recommended structure for Power BI. Because there are fewer "hops" between tables, performance is faster, and the DAX formulas are simpler to write.

Snowflake Schema: * Structure: An extension of the Star schema where Dimension tables are "normalized" (broken down) into further sub-dimension tables (e.g., Product connects to Sub-Category, which connects to Category).

Significance: While it saves disk space by reducing redundant data, it makes the model more complex. In Power BI, this can slow down report performance because the engine has to work harder to navigate the relationships.

2. Primary Key vs. Foreign Key
These keys are the "hooks" that allow two tables to talk to each other.

Primary Key (PK): * A unique identifier for every row in a table. For example, in a Customers table, the CustomerID is the Primary Key. It cannot have duplicates and cannot be blank.

Foreign Key (FK): * A column in one table that points to the Primary Key in another table. For example, in the Sales table, the CustomerID column is a Foreign Key used to link a specific sale back to a specific person in the Customers table.

3. Why is Cardinality Important?
Cardinality defines the nature of the relationship between two tables based on how many times a key appears. It is crucial because it dictates how data filters flow through your report.

One-to-Many (1:*) – The Most Common: One row in the Dimension table (e.g., one unique Product ID) matches many rows in the Fact table (e.g., many sales of that product). This is the healthiest relationship for Power BI.

One-to-One (1:1): Both tables share the same unique keys. Usually, these should just be merged into one table.

Many-to-Many (:): Multiple rows in both tables share the same keys.

Why it's risky: This can lead to "ambiguous" results where Power BI isn't sure how to calculate totals, often resulting in inaccurate data or performance lag.

Pro Tip: Always strive for One-to-Many relationships in a Star Schema. It keeps your model "lean and mean," ensuring your dashboards snap to attention the moment a user clicks a filter.

Q5.  What is the difference between a report and a dashboard in Power BI?

ANS.  In Power BI, the primary difference is their purpose: reports are for deep-dive analysis and exploration, while dashboards are single-page summaries for high-level monitoring at a glance

1. Structure and ScopeA report is a multi-perspective view of a specific dataset. It can have multiple pages and provides detailed storytelling through various visuals. A dashboard acts as a "pinboard," aggregating the most important visuals (called tiles) from one or more reports into a single-page overview.

2. Analytical Depth vs. SnapshotReports allow for thorough investigation using advanced features like DAX expressions, cross-filtering, and drill-through pages to understand the "why" behind data trends. Dashboards are designed for rapid assessment, helping users know "how" things stand in five seconds or less without complex manipulation

3. Creation and AccessYou can build reports in Power BI Desktop and then publish them. However, you cannot create dashboards in the Desktop application; they must be created in the Power BI Service by pinning visuals from published reports.

4. Special FunctionalityAlerts: Dashboards allow you to set email alerts when a specific metric (like a KPI) crosses a threshold; reports do not support this.Navigation: Clicking any tile on a dashboard usually navigates the user directly to the source report page for more detail
